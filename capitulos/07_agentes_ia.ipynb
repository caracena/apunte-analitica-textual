## 7.1 ¿Qué es un agente?

Un agente de IA combina al menos cuatro piezas:

1. un modelo de lenguaje;
2. un conjunto de herramientas disponibles;
3. una memoria o estado;
4. una política para decidir el siguiente paso.

En términos simples, un agente observa, decide, actúa y vuelve a evaluar.

## 7.2 Patrón básico de arquitectura


```text
Usuario -> Modelo -> Selección de herramienta -> Ejecución -> Memoria -> Respuesta
```


Ejemplos de herramientas:

- buscador documental;
- base de datos;
- calculadora;
- sistema de tickets;
- agenda o correo.

## 7.3 Ejemplo simple en Python

El siguiente ejemplo ilustra la idea sin depender de un proveedor específico:


````python
herramientas = {
    "buscar_politica": lambda consulta: f"Resultado documental para: {consulta}",
    "crear_ticket": lambda detalle: f"Ticket creado con detalle: {detalle}",
}

memoria = []

def agente(regla, entrada_usuario):
    memoria.append({"usuario": entrada_usuario})

    if "reembolso" in entrada_usuario.lower():
        evidencia = herramientas["buscar_politica"]("política de reembolsos")
        accion = herramientas["crear_ticket"](entrada_usuario)
        respuesta = f"{evidencia}\n{accion}"
    else:
        respuesta = "No se activó ninguna herramienta específica."

    memoria.append({"agente": respuesta})
    return respuesta

print(agente("soporte", "Necesito un reembolso por cobro duplicado"))
````

No es un agente autónomo completo, pero muestra la lógica de orquestación.

## 7.4 Memoria

La memoria puede ser:

- de corto plazo: historial de la conversación actual;
- de largo plazo: preferencias, tareas previas, documentos o contexto persistente.

Hay que decidir con cuidado qué recordar, durante cuánto tiempo y con qué controles de privacidad.

## 7.5 Evaluación de agentes

Un agente no se evalúa solo por sonar bien. Algunas preguntas más útiles:

- ¿usa la herramienta correcta?
- ¿invoca pasos innecesarios?
- ¿respeta reglas del negocio?
- ¿maneja errores y ambigüedad?
- ¿deja trazabilidad?

## 7.6 Riesgos frecuentes

- ejecución de herramientas sobre datos equivocados;
- loops innecesarios;
- sobreconfianza del modelo;
- acceso indebido a información sensible;
- falta de auditoría.

## 7.7 Buenas prácticas

1. Diseñar herramientas pequeñas y bien definidas.
2. Registrar entradas, salidas y decisiones.
3. Limitar acciones de alto impacto.
4. Probar con casos adversariales.
5. Mantener siempre una vía de escalamiento humano.
