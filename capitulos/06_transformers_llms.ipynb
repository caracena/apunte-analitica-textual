## 6.1 Del embedding estático al contexto

En Word2Vec, la palabra `banco` tiene un solo vector, aunque en distintos contextos pueda significar institución financiera o asiento. Los Transformers producen representaciones contextuales: el vector depende de las palabras vecinas.

## 6.2 La idea central de Transformer

La arquitectura Transformer se basa en mecanismos de atención. En vez de procesar palabra por palabra como una RNN, calcula qué tokens deben atenderse mutuamente para construir una representación más informativa.

Conceptos clave:

- tokenización;
- embeddings de entrada;
- atención multi-cabeza;
- capas feed-forward;
- codificación posicional.

## 6.3 Pre-entrenamiento y fine-tuning

Primero, el modelo aprende sobre grandes corpus generales. Luego se adapta a una tarea específica:

- clasificación;
- reconocimiento de entidades;
- resumen;
- respuesta a preguntas.

Eso reduce la necesidad de partir desde cero.

## 6.5 LLMs

Los Large Language Models extienden esta lógica a escalas mucho mayores. Además de clasificación, pueden:

- generar texto;
- resumir documentos;
- reformular mensajes;
- extraer información en formato estructurado;
- asistir en tareas conversacionales.

## 6.6 Patrones de uso

Tres formas comunes de aprovechar un LLM:

1. prompting directo;
2. RAG para recuperar contexto externo;
3. fine-tuning o adaptación.

## 6.7 Cuándo usar un LLM

Úsalo cuando el problema requiera:

- comprensión contextual alta;
- generación flexible;
- adaptación rápida a tareas diversas;
- interacción conversacional.

No es necesariamente la mejor opción si:

- basta con una clasificación simple y estable;
- necesitas latencia mínima y costos bajos;
- la explicabilidad es prioritaria;
- el flujo está muy regulado y tolera poca variación.

## 6.8 Riesgos y gobernanza

Los LLMs introducen riesgos adicionales:

- alucinaciones;
- fuga de datos sensibles;
- respuestas no deterministas;
- costos variables;
- dependencia de proveedores.

Por eso el diseño de producto debe incluir evaluación, guardrails y monitoreo.

## 6.9 Ejemplo de resumen con pipeline


```python
from transformers import pipeline

resumidor = pipeline("summarization", model="facebook/bart-large-cnn")
texto = """
La analítica textual combina técnicas estadísticas, lingüísticas y de aprendizaje automático
para transformar documentos en información útil para la toma de decisiones.
"""

resumidor(texto, max_length=40, min_length=15, do_sample=False)
```

Este ejemplo es ilustrativo. En producción deberíamos validar idioma, longitud, costo y consistencia.
