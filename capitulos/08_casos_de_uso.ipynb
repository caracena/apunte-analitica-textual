## 8.2 Legaltech

Problemas comunes:

- clasificación de escritos;
- búsqueda semántica de jurisprudencia;
- extracción de cláusulas;
- detección de riesgos contractuales.

Técnicas útiles:

- regex para campos estructurados;
- TF-IDF + similitud para recuperación inicial;
- embeddings y RAG para búsqueda semántica;
- LLMs para resumen asistido con revisión humana.

Métrica crítica:

- `recall` alto cuando omitir un antecedente es costoso.

## 8.3 Salud

Problemas comunes:

- clasificación de notas clínicas;
- normalización de terminología;
- detección de eventos adversos;
- resumen de fichas extensas.

Restricciones especiales:

- privacidad;
- trazabilidad;
- validación clínica;
- sesgos poblacionales.

En salud, muchas veces conviene combinar extracción estructurada, modelos supervisados y supervisión humana.

## 8.4 Finanzas

Problemas comunes:

- monitoreo de noticias;
- análisis de reportes y memorias;
- clasificación de reclamos;
- detección de fraude documental.

Factores críticos:

- velocidad;
- auditabilidad;
- robustez frente a cambios regulatorios;
- costo de falsos positivos y falsos negativos.

## 8.5 Atención al cliente

Problemas comunes:

- ruteo automático de tickets;
- análisis de sentimiento;
- resumen de conversaciones;
- sugerencia de respuesta.

Aquí suelen convivir muy bien enfoques clásicos y modernos:

- clasificación rápida con TF-IDF + SVM;
- resumen con LLM;
- agente para consultar políticas y crear tickets.

## 8.6 Mini caso integrado

Supongamos un centro de soporte que recibe mensajes de clientes. Una solución razonable podría ser:

1. limpiar texto y metadatos;
2. clasificar tipo de solicitud;
3. detectar urgencia;
4. resumir el caso;
5. derivar a un agente con herramientas internas.

Ejemplo base:


```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC

textos = [
    "me cobraron dos veces la misma suscripción",
    "no puedo ingresar a mi cuenta",
    "quiero cambiar el plan contratado",
    "necesito cerrar mi cuenta de inmediato"
]

etiquetas = ["facturación", "acceso", "comercial", "baja"]

modelo = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("clf", LinearSVC())
])

modelo.fit(textos, etiquetas)
modelo.predict(["tengo un cobro duplicado en mi tarjeta"])
```

Este clasificador puede ser solo la primera etapa de un flujo más amplio.

## 8.7 Criterios para seleccionar solución

Usa técnicas clásicas cuando:

- el problema está bien definido;
- hay etiquetas claras;
- necesitas explicabilidad;
- el costo y la latencia son sensibles.

Usa modelos modernos o LLMs cuando:

- la tarea exige comprensión contextual compleja;
- necesitas generación o reformulación;
- el espacio de tareas cambia rápido;
- puedes invertir en evaluación y gobernanza.
