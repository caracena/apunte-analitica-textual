## Objetivos

Los modelos vistos hasta ahora tratan palabras como dimensiones casi independientes. Los embeddings introducen una idea distinta: representar palabras como vectores densos que capturan contexto y cercanía semántica.

## 5.1 Intuición

La hipótesis distribucional dice: palabras que aparecen en contextos similares tienden a tener significados relacionados. Word2Vec aprende vectores para que esa intuición quede codificada en el espacio.

Ejemplos esperados:

- `doctor` cerca de `médico`;
- `banco` cerca de `finanzas` en un corpus económico;
- `tribunal` cerca de `juez` en un corpus legal.

## 5.2 Corpus pequeño de entrenamiento


```python
corpus = [
    "el banco aprueba créditos para vivienda".split(),
    "la fintech ofrece pagos digitales y créditos".split(),
    "el hospital coordina atención médica y urgencias".split(),
    "la clínica mejora la atención de pacientes".split(),
    "el tribunal revisa contratos y evidencia".split(),
    "el abogado analiza cláusulas del contrato".split()
]
```

## 5.3 Entrenamiento con Word2Vec


```python
from gensim.models import Word2Vec

modelo_w2v = Word2Vec(
    sentences=corpus,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    sg=1,
    epochs=200
)
```

Parámetros importantes:

- `vector_size`: tamaño del embedding;
- `window`: amplitud del contexto;
- `min_count`: frecuencia mínima;
- `sg=1`: usa Skip-Gram en vez de CBOW.

## 5.4 Consultar palabras similares


```python
modelo_w2v.wv.most_similar("créditos", topn=5)
```

Si el corpus es pequeño, los resultados serán inestables. Aun así sirven para comprender la lógica del método.

## 5.5 Inspección de similitud


```python
sim_banco_fintech = modelo_w2v.wv.similarity("banco", "fintech")
sim_hospital_clínica = modelo_w2v.wv.similarity("hospital", "clínica")

print(round(sim_banco_fintech, 3))
print(round(sim_hospital_clínica, 3))
```

## 5.6 ¿Qué ganamos con embeddings?

Frente a TF-IDF, los embeddings:

- reducen dimensionalidad;
- capturan cierta cercanía semántica;
- permiten usar promedios o secuencias como entrada a modelos posteriores.

## 5.7 Limitaciones

Word2Vec clásico tiene límites importantes:

- produce un vector fijo por palabra, sin distinguir contexto;
- puede amplificar sesgos del corpus;
- necesita suficiente volumen y calidad de datos;
- no maneja bien palabras no vistas.

Por eso los embeddings contextuales de Transformers fueron un salto tan importante.

## 5.8 Uso práctico

Los embeddings sirven para:

- expandir búsquedas con términos parecidos;
- medir similitud semántica;
- inicializar modelos posteriores;
- explorar vocabulario especializado.
