## 1.1 ¿Qué es la Analítica Textual?

La **Analítica Textual** (también llamada *Text Analytics* o *Text Mining*) es el proceso de extraer información significativa y patrones a partir de datos textuales no estructurados. A diferencia de los datos tabulares, el texto requiere técnicas especializadas para su procesamiento y análisis.

### El volumen de datos textuales

Se estima que más del **80% de los datos empresariales** son no estructurados, y una gran proporción de estos son textos: correos electrónicos, contratos, reportes, publicaciones en redes sociales, tickets de soporte, entre otros.

### Pipeline típico de Analítica Textual

```
Texto crudo → Limpieza → Representación numérica → Modelado → Interpretación
```

## 1.3 Retos del Lenguaje Natural

El lenguaje natural presenta desafíos únicos para las máquinas:

- **Ambigüedad léxica**: Una misma palabra puede tener múltiples significados (*banco* → entidad financiera o asiento).
- **Ambigüedad sintáctica**: Una oración puede interpretarse de varias formas.
- **Variabilidad**: Sinónimos, jerga, errores ortográficos, abreviaciones.
- **Contexto**: El significado depende del contexto en que se usa una palabra.
- **Idioma y cultura**: Expresiones idiomáticas, sarcasmo, ironía.

## 1.4 Limpieza de datos textuales

Antes de cualquier análisis, es fundamental limpiar y normalizar el texto. Los pasos típicos incluyen:

1. Convertir a minúsculas
2. Eliminar caracteres especiales y puntuación
3. Eliminar números (según el caso)
4. Eliminar *stopwords* (palabras vacías)
5. Aplicar *stemming* o *lemmatization*

````python
# Ejemplo básico de limpieza de texto
texto_crudo = "¡Hola! Este es un EJEMPLO de texto con números 123 y caracteres especiales @#$%."

# Paso 1: Convertir a minúsculas
texto = texto_crudo.lower()
print("Minúsculas:", texto)
````

## 1.5 Expresiones Regulares (Regex)

Las **expresiones regulares** son secuencias de caracteres que definen un patrón de búsqueda. Son una herramienta fundamental para la limpieza y extracción de información en textos.

### Metacaracteres principales

| Patrón | Descripción | Ejemplo |
|--------|-------------|---------|
| `.` | Cualquier carácter | `a.c` → "abc", "a1c" |
| `\d` | Dígito (0-9) | `\d{3}` → "123" |
| `\w` | Carácter alfanumérico | `\w+` → "hola" |
| `\s` | Espacio en blanco | `\s+` → " " |
| `^` | Inicio de línea | `^Hola` → "Hola..." |
| `$` | Fin de línea | `fin$` → "...fin" |
| `*` | 0 o más repeticiones | `ab*` → "a", "ab", "abb" |
| `+` | 1 o más repeticiones | `ab+` → "ab", "abb" |
| `?` | 0 o 1 repetición | `colou?r` → "color", "colour" |
| `[]` | Conjunto de caracteres | `[aeiou]` → vocales |
| `\|` | Alternativa (OR) | `gato\|perro` |

````python
import re

regex = "a"
string_para_buscar = "Santiago"
print(re.findall(regex, string_para_buscar))

regex = "u"
string_para_buscar = "Santiago"
print(re.findall(regex, string_para_buscar))  # []

regex = "San"
string_para_buscar = "Santiago"
print(re.findall(regex, string_para_buscar))  # ['San']

print(re.findall("tia", "Santiago"))  # ['tia']
print(re.findall("san", "Santiago"))  # []

print(re.findall(".", "Santiago"))       # ['S', 'a', 'n', 't', 'i', 'a', 'g', 'o']
print(re.findall(".", "a_94j-+' 1H"))   # ['a', '_', '9', '4', 'j', '-', '+', "'", ' ', '1', 'H']
print(re.findall(".", ""))               # []
````

````python
print([re.findall("a.b", s) for s in ["aaa", "aab", "abb", "acadb"]])
````

````python
print(re.findall("a+", "Santiago"))  # ['a', 'a']
print(re.findall("x*", "Santiago"))  # ['', '', '', '', '', '', '', '', '']
````

````python
print(re.findall("[aeiu]", "Concon"))      # []
print(re.findall("[aeiou]", "Concon"))     # ['o', 'o']
print(re.findall("[^aeiou]", "bcdfghijk")) # ['b', 'c', 'd', 'f', 'g', 'h', 'j', 'k']
````

````python
print(re.findall("[a-m]", "xyz"))       # []
print(re.findall("[a-m]", "ABC"))       # []
print(re.findall("[a-mA-M]", "ABC"))    # ['A', 'B', 'C']
````

````python
print([re.findall("^a.+", s) for s in ["bab", "aab"]])
````

````python
print([re.findall("b$", s) for s in ["bab", "aab"]])
print([re.findall("^[ab]+$", s) for s in ["bab", "aab", "abc"]])
````

````python
import re

texto = "Contactar a juan.perez@email.com o maria_lopez@empresa.cl para más info. Tel: +56 9 1234 5678"

# Extraer correos electrónicos
emails = re.findall(r'[\w.+-]+@[\w-]+\.[\w.-]+', texto)
print("Emails encontrados:", emails)

# Extraer números de teléfono
telefonos = re.findall(r'\+?\d[\d\s]{7,}\d', texto)
print("Teléfonos encontrados:", telefonos)
````

````python
# Limpieza de texto con Regex
texto_sucio = """<p>Este es un texto con HTML</p>
URLs: https://www.ejemplo.com y http://otro-sitio.org
Menciones: @usuario1 y @usuario2
Hashtags: #analitica #textual #NLP
Emojis y símbolos: 😀 ★ © ® ™
Números: El precio es $1.500.000 CLP"""

def limpiar_texto(texto):
    """Pipeline de limpieza de texto usando regex."""
    # Eliminar etiquetas HTML
    texto = re.sub(r'<[^>]+>', '', texto)
    # Eliminar URLs
    texto = re.sub(r'https?://\S+', '', texto)
    # Eliminar menciones
    texto = re.sub(r'@\w+', '', texto)
    # Eliminar hashtags
    texto = re.sub(r'#\w+', '', texto)
    # Eliminar caracteres no alfanuméricos (excepto espacios)
    texto = re.sub(r'[^\w\sáéíóúñü]', '', texto)
    # Eliminar espacios múltiples
    texto = re.sub(r'\s+', ' ', texto).strip()
    # Convertir a minúsculas
    texto = texto.lower()
    return texto

texto_limpio = limpiar_texto(texto_sucio)
print("Texto limpio:")
print(texto_limpio)
````

## 1.6 Tokenización

La **tokenización** es el proceso de dividir un texto en unidades más pequeñas llamadas *tokens* (generalmente palabras o subpalabras).

````python
import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords

texto = "La analítica textual permite extraer información valiosa. Es una disciplina en constante evolución."

# Tokenización por oraciones
oraciones = sent_tokenize(texto, language='spanish')
print("Oraciones:", oraciones)

# Tokenización por palabras
palabras = word_tokenize(texto, language='spanish')
print("Palabras:", palabras)

# Eliminar stopwords
stop_words = set(stopwords.words('spanish'))
palabras_filtradas = [p for p in palabras if p.lower() not in stop_words and p.isalpha()]
print("Sin stopwords:", palabras_filtradas)
````

## 1.7 Stemming y Lematización

- **Stemming**: Reduce las palabras a su raíz cortando sufijos (más agresivo, puede generar raíces no válidas).
- **Lematización**: Reduce las palabras a su forma base o *lema* usando un diccionario (más preciso).

````python
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer('spanish')

palabras = ["corriendo", "correr", "corrió", "computadoras", "computación", "análisis", "analizar"]

print("Stemming:")
for palabra in palabras:
    print(f"  {palabra} → {stemmer.stem(palabra)}")
````

### Lematización con spaCy

A diferencia del stemming, la **lematización** utiliza un modelo lingüístico que conoce la morfología del idioma, por lo que siempre devuelve palabras reales (lemas). La biblioteca [spaCy](https://spacy.io/) ofrece modelos preentrenados para español que incluyen lematización, etiquetado gramatical (POS tagging) y más.

````python
# Instalar spaCy y descargar el modelo en español (ejecutar solo una vez)
!pip install -q spacy
!python -m spacy download es_core_news_sm -q
````

````python
import spacy

# Cargar el modelo en español
nlp = spacy.load('es_core_news_sm')

# Mismas palabras usadas en el ejemplo de stemming
palabras = ["corriendo", "correr", "corrió", "computadoras", "computación", "análisis", "analizar"]

print("Lematización con spaCy:")
for palabra in palabras:
    doc = nlp(palabra)
    lema = doc[0].lemma_
    print(f"  {palabra} → {lema}")

# Comparación lado a lado: stemming vs. lematización
print("\n--- Comparación: Stemming vs. Lematización ---")
print(f"{'Palabra':<16} {'Stemming':<16} {'Lema (spaCy)':<16}")
print("-" * 48)
for palabra in palabras:
    stem = stemmer.stem(palabra)
    lema = nlp(palabra)[0].lemma_
    print(f"{palabra:<16} {stem:<16} {lema:<16}")
````

## 1.8 Ejemplo integrador: Limpieza de un corpus

Construyamos un pipeline completo de limpieza aplicado a un pequeño corpus de documentos.

````python
corpus = [
    "El machine learning está REVOLUCIONANDO la industria financiera!!! 🚀",
    "<b>Nuevo</b> modelo de NLP con 99.5% de precisión - ver https://ejemplo.com",
    "@dr_garcia presenta resultados del estudio #SaludDigital en el congreso 2024.",
    "Los contratos inteligentes (smart contracts) simplifican procesos de la industria legal.",
    "El chatbot resolvió 1.500 consultas en enero, un 30% más que dic. anterior, según estudio"
]

def pipeline_limpieza(texto):
    """Pipeline completo de preprocesamiento de texto."""
    # 1. Limpieza con regex
    texto = re.sub(r'<[^>]+>', '', texto)          # HTML
    texto = re.sub(r'https?://\S+', '', texto)     # URLs
    texto = re.sub(r'@\w+', '', texto)             # Menciones
    texto = re.sub(r'#\w+', '', texto)             # Hashtags
    texto = re.sub(r'[^\w\sáéíóúñü]', '', texto)  # Caracteres especiales
    texto = re.sub(r'\d+', '', texto)              # Números
    texto = re.sub(r'\s+', ' ', texto).strip()     # Espacios múltiples
    texto = texto.lower()                           # Minúsculas
    
    # 2. Tokenización
    tokens = word_tokenize(texto, language='spanish')
    
    # 3. Eliminar stopwords
    tokens = [t for t in tokens if t not in stop_words and len(t) > 2]
    
    # 4. Stemming
    tokens = [stemmer.stem(t) for t in tokens]
    
    return tokens

print("Corpus procesado:")
for i, doc in enumerate(corpus):
    tokens = pipeline_limpieza(doc)
    print(f"  Doc {i+1}: {tokens}")
````

## 1.9 Vocabulario y frecuencia de palabras

Una vez procesado el corpus, es útil analizar qué palabras (o stems) aparecen y con qué frecuencia. El **vocabulario** es el conjunto de tokens únicos del corpus. La distribución de frecuencias nos permite identificar los términos más relevantes y verificar que nuestra limpieza fue adecuada.

````python
from collections import Counter

# Procesar todo el corpus y juntar los tokens
todos_los_tokens = []
for doc in corpus:
    tokens = pipeline_limpieza(doc)
    todos_los_tokens.extend(tokens)

# Calcular el vocabulario (tokens únicos)
vocabulario = sorted(set(todos_los_tokens))
print(f"Total de tokens en el corpus: {len(todos_los_tokens)}")
print(f"Tamaño del vocabulario (tokens únicos): {len(vocabulario)}")
print(f"\nVocabulario:\n{vocabulario}")

# Calcular frecuencias
frecuencias = Counter(todos_los_tokens)
print(f"\nTop 10 tokens más frecuentes:")
for token, freq in frecuencias.most_common(10):
    print(f"  {token}: {freq}")
````

````python
import matplotlib.pyplot as plt

# Obtener los tokens más frecuentes para graficar
top_n = 15
tokens_top = frecuencias.most_common(top_n)
palabras = [t[0] for t in tokens_top]
conteos = [t[1] for t in tokens_top]

# Gráfica de barras horizontales
plt.figure(figsize=(10, 6))
plt.barh(range(len(palabras)), conteos, color='steelblue')
plt.yticks(range(len(palabras)), palabras)
plt.xlabel('Frecuencia')
plt.ylabel('Token (stem)')
plt.title(f'Top {top_n} tokens más frecuentes en el corpus')
plt.gca().invert_yaxis()  # El más frecuente arriba
plt.tight_layout()
plt.show()
````
